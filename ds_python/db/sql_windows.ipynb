#### table_schema

```python
%%sql result_set << 
SELECT table_name, table_schema, table_type
FROM information_schema.tables
WHERE table_schema = 'Chinook'
ORDER BY table_name ASC;
```

```python
%%sql result_set <<
SELECT * FROM
Album LEFT JSON Artist
ON Album.ArtistId = Artist.ArtistId;
```

```python
%%sql result_set << 
SELECT
  Name
  , AlbumId
  , Composer
  , Milliseconds
  , Bytes
  -- , MediaTypeId
  , COUNT(Composer) AS composer_num
  -- AVG(Bytes) OVER (PARTITION BY MediaTypeId)
FROM Track
WHERE MediaTypeId = 1
-- AND composer_num > 10
GROUP BY Composer
ORDER BY composer_num DESC;
```
